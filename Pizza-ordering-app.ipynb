```python
import tkinter as tk
from tkinter import*
from tkinter import messagebox
from tkinter import ttk
```

```python
class Pizza:
    def __init__(self, name, email, number, address, type, size, toppings, num_pizza):
        self.name = name
        self.email = email
        self.number = number
        self.address = address
        self.type = type
        self.size = size
        self.toppings = toppings
        self.num_pizza = num_pizza

    def order(self):
        return f"Your order sent, It will arrive soon!"
    
def place_order():
    name = name_entry.get()
    email = email_entry.get()
    number = number_entry.get()
    address = address_entry.get()
    type = type_var.get()
    size = size_var.get()
    size = size_entry.get()
    toppings = toppings_entry.get()
    num_pizza = num_pizza_entry
    if name and email and number and address and type and size and toppings and num_pizza:
        pizza = Pizza(name, email, number, address, type, size, toppings, num_pizza)
        messagebox.showinfo("Order", pizza.order())
    else:
        messagebox.showwarning("Missing Info","Please enter All Info!")
def order_cancel():
    messagebox.askokcancel("Cancel Order","Do you want to cancel your order?")

app = tk.Tk()
app.title("Pizza Ordering App")
app.configure(bg="#000000")
app.geometry("880x600")

def submit():
    toppings = toppings_entry.get()
    try:
        if not all (char.isalpha() or char.isspace() for char in toppings):
            raise ValueError("Only alphabets and spaces allowed!")
        messagebox.showinfo(f"Toppings all done!")
    except Exception as e:
        messagebox.showinfo("Invalid Info",str(e))

lab1 = tk.Label(app, text="🍕🍕Pizza Ordering App🍕🍕", bg="#DAA520", font=("bold",22,"italic"))
lab1.grid(row=1, columnspan=5, column=1)

name_lab= tk.Label(app, text="Name", font=("italic",14,"bold"), bg="#000000", fg="#DAA520", pady=6)
name_lab.grid(row=2,column=0,padx=5,pady=5,sticky="w")

name = StringVar()
name_box = name_entry = tk.Entry(app, bd=5, textvariable=name)
name_box.grid(row=2,column=1,padx=5,pady=5)

email_lab= tk.Label(app, text="Email", font=("italic",12,"bold"), bg="#000000", fg="#DAA520", pady=6)
email_lab.grid(row=2,column=2, padx=5, pady=5, sticky="e")

email = StringVar()
email_box = email_entry = tk.Entry(app, bd=5, textvariable=email)
email_box.grid(row=2,column=3)

number_lab= tk.Label(app, text="Phone", font=("italic",12,"bold"), bg="#000000", fg="#DAA520", pady=6)
number_lab.grid(row=3,column=0, padx=5, pady=5, sticky="w")

number = StringVar()
number_box = number_entry = tk.Entry(app, bd=5, textvariable=number)
number_box.grid(row=3,column=1)

address_lab= tk.Label(app, text="Address", font=("italic",12,"bold"), bg="#000000", fg="#DAA520", pady=6)
address_lab.grid(row=3,column=2, padx=5, pady=5, sticky="e")

address = StringVar()
address_box = address_entry = tk.Entry(app, bd=5, textvariable=address)
address_box.grid(row=3,column=3)

lab2 = tk.Label(app, text="🍕Choose Type🍕", bg="#DAA520", font=("bold",16,"italic"))
lab2.grid(row=4, columnspan=4, column=1)

type_var = tk.StringVar(value="Normal Pizza")
b1 = tk.Checkbutton(app, text="Normal Pizza", bg ="#000000", fg="#DAA520",  font=("bold",10,"italic"), variable=type_var)
b1.grid(row=5, column=1, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="Crown Crust Pizza")
b2 = tk.Checkbutton(app, text="Crown Crust Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b2.grid(row=5, column=2, padx=5, pady=5, sticky="n")

type_var = tk.StringVar(value="Tikka Kebab Pizza")
b3 = tk.Checkbutton(app, text="Tikka Kebab Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b3.grid(row=5, column=3, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="Chessy Pizza")
b4 = tk.Checkbutton(app, text="Cheesy Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b4.grid(row=6, column=1, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="Pepperoni Pizza")
b5 = tk.Checkbutton(app, text="Pepperoni Pizza", bg="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b5.grid(row=6, column=2, padx=5, pady=5, sticky="n")

type_var = tk.StringVar(value="Chicken tender Pizza")
b6 = tk.Checkbutton(app, text="Chicken tender Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b6.grid(row=6, column=3, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="Lasagna Pizza")
b7 = tk.Checkbutton(app, text="Lasagna Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b7.grid(row=7, column=1, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="Hot'n'Spicy Pizza")
b8 = tk.Checkbutton(app, text="Hot'n'spicy Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b8.grid(row=7, column=2, padx=5, pady=5, stick="n")

type_var = tk.StringVar(value="Chicken Fajita Pizza")
b9 = tk.Checkbutton(app, text="Chicken Fajita Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b9.grid(row=7, column=3, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="Behari Boti Pizza")
b10 = tk.Checkbutton(app, text="Behari Boti Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b10.grid(row=8, column=1, padx=5, pady=5, sticky="w")

type_var = tk.StringVar(value="King Stuffer Pizza")
b11 = tk.Checkbutton(app, text="King Stuffer Pizza", bg ="#000000", fg="#DAA520", font=("bold",10,"italic"), variable=type_var)
b11.grid(row=8, column=2, padx=5, pady=5, sticky="n")

type_var = tk.StringVar(value="Grilled Pizza")
b12= tk.Checkbutton(app, text="Grilled  Pizza", bg ="#000000", fg="#DAA520",  font=("bold",10,"italic"), variable=type_var)
b12.grid(row=8, column=3, padx=5, pady=5, sticky="w")

lab3 = tk.Label(app, text="🍕Choose size🍕", bg="#DAA520", font=("bold",16,"italic"))
lab3.grid(row=9, columnspan=4, column=1)

size_var = tk.StringVar()
b13 = tk.Radiobutton(app, text="Small", bg ="#000000", fg="#DAA520", font=("bold",13,"italic"), variable=size_var, value="Small")
b13.grid(row=10, column=1)

b14 = tk.Radiobutton(app, text="Medium", bg ="#000000", fg="#DAA520",  font=("bold",13,"italic"), variable=size_var, value="Medium")
b14.grid(row=10, column=2)

b15 = tk.Radiobutton(app, text="Large", bg ="#000000", fg="#DAA520", font=("bold",13,"italic"), variable=size_var, value="Large")
b15.grid(row=10, column=3)

size_box = size_entry = tk.Entry(app, bd=5)
size_box.grid(row=11, column=2)

labsize = tk.Label(app, text="(😊Entry box is for those who want to order more than 1 pizza, Thank You!😊)", bg="#000000", fg="#DAA520", font=("bold",16,"italic"))
labsize.grid(row=12, columnspan=1, column=2)

lab4 = tk.Label(app, text="🍕Choose Custom Toppings🍕", bg="#DAA520", font=("bold",14,"italic"))
lab4.grid(row=13, columnspan=4, column=1)

top_box = toppings_entry = tk.Entry(app, bd=5)
top_box.grid(row=14, column=2)

order_b = tk.Button(app, text="✔Place Order✔", bg="Green",font=("bold",13,"italic"), command=place_order, bd=7)
order_b.grid(row=20, column=1)

cancel_b = tk.Button(app, text="❌Cancel Order❌", bg="Red", font=("bold",13,"italic"), command=order_cancel, bd=7)
cancel_b.grid(row=20, column=3)

lab5 = tk.Label(app, text="😊For Original Toppings, Write: All Original Toppings😊", bg="#000000",fg="#DAA520", font=("bold",14))
lab5.grid(row=15, column=2)

lab6 = tk.Label(app, text="🥤For cold drinks, Use this🥤", bg="#DAA520", font=("bold",14))
lab6.grid(row=18, column=2)

drinks_type = tk.IntVar()
drinks_choosen = ttk.Combobox(app, width=12,textvariable=drinks_type)
drinks_choosen["value"]=["half ltr Next Cola","1 ltr Next Cola","1.5 ltr Next Cola","half ltr 7UP","1 ltr 7UP","1.5 ltr 7UP","half ltr Mirinda","1 ltr Mirinda","1.5 ltr Mirinda","half ltr Water","1 ltr Water"]
drinks_choosen.grid(row=19, columnspan=2, column=1)

drink_box = size_entry = tk.Entry(app, bd=5)
drink_box.grid(row=19, columnspan=2, column=2)

lab7 = tk.Label(app, text="🍕Select the amounts of Pizza🍕", bg="#DAA520", font=("bold",14))
lab7.grid(row=16, column=2)

num_pizza_box = num_pizza_entry = tk.Entry(app, bd=5)
num_pizza_box.grid(row=17, column=2)

app.mainloop()
```
